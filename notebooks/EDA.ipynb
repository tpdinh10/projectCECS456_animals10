# EDA – Animals10 Dataset (Google Images)

This notebook explores the custom Animals10 dataset located in `data/animals10/`.
Each subfolder in `data/animals10/` corresponds to one animal class.

```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

DATA_DIR = os.path.join("..", "data", "animals10")

classes = sorted(
    d for d in os.listdir(DATA_DIR)
    if os.path.isdir(os.path.join(DATA_DIR, d))
)

print("Classes:", classes)
```

```python
# Count number of images per class

counts = {}
for cls in classes:
    folder = os.path.join(DATA_DIR, cls)
    files = [
        f for f in os.listdir(folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp"))
    ]
    counts[cls] = len(files)

counts
```

```python
# Plot class distribution as a bar chart

plt.figure(figsize=(10, 4))
plt.bar(counts.keys(), counts.values())
plt.xticks(rotation=45)
plt.ylabel("Number of images")
plt.title("Class distribution")
plt.tight_layout()
plt.show()
```

```python
# Show a few sample images from each class 

n_samples_per_class = 3

for cls in classes:
    folder = os.path.join(DATA_DIR, cls)
    files = [
        f for f in os.listdir(folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp"))
    ]
    print(f"Class: {cls}, total images: {len(files)}")

    plt.figure(figsize=(n_samples_per_class * 2, 2))
    for i, fname in enumerate(files[:n_samples_per_class]):
        img_path = os.path.join(folder, fname)
        img = mpimg.imread(img_path)
        plt.subplot(1, n_samples_per_class, i + 1)
        plt.imshow(img)
        plt.axis("off")
    plt.suptitle(f"Samples from class: {cls}")
    plt.show()
```
