# Training Results – Animals10 CNN

This notebook visualizes the training history and evaluation results produced by `src/train.py`. It expects files in the `results/` folder:

- `history.json`
- `training_curves.png`
- `confusion_matrix.png`

```python
import os
import json
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

RESULTS_DIR = os.path.join("..", "results")

history_path = os.path.join(RESULTS_DIR, "history.json")

if not os.path.exists(history_path):
    raise FileNotFoundError(
        f"{history_path} not found. Run src/train.py first to generate training history."
    )

with open(history_path, "r", encoding="utf-8") as f:
    history = json.load(f)

history.keys()
```

```python
# Plot accuracy curves from loaded history

acc = history.get("accuracy", [])
val_acc = history.get("val_accuracy", [])

plt.figure(figsize=(6, 4))
plt.plot(acc, label="train acc")
plt.plot(val_acc, label="val acc")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Epoch")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Plot loss curves from loaded history

loss = history.get("loss", [])
val_loss = history.get("val_loss", [])

plt.figure(figsize=(6, 4))
plt.plot(loss, label="train loss")
plt.plot(val_loss, label="val loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss vs Epoch")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Display saved training_curves.png 

curves_path = os.path.join(RESULTS_DIR, "training_curves.png")
if os.path.exists(curves_path):
    img = mpimg.imread(curves_path)
    plt.figure(figsize=(8, 4))
    plt.imshow(img)
    plt.axis("off")
    plt.title("Training curves (from train.py)")
    plt.show()
else:
    print(f"{curves_path} not found.")
```

```python
# Display confusion_matrix.png 

cm_path = os.path.join(RESULTS_DIR, "confusion_matrix.png")
if os.path.exists(cm_path):
    img = mpimg.imread(cm_path)
    plt.figure(figsize=(6, 6))
    plt.imshow(img)
    plt.axis("off")
    plt.title("Confusion Matrix (from train.py)")
    plt.show()
else:
    print(f"{cm_path} not found.")
```
